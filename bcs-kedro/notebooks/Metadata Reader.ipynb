```python
import re
import os
import pandas as pd
from itertools import repeat
from tqdm.notebook import tqdm
from PIL import Image
```

```python
BASE_DIR = '../data/02_intermediate/spatial-attention-pcr'
```

```python
os.listdir(os.path.join(BASE_DIR, 'pcr'))
```

```python
metadata_list = []

for root, dirs, files in os.walk(BASE_DIR):
    root_abspath = os.path.abspath(root)
    for i, file in enumerate(files):
        print(f'File({i+1} of {len(files)})')
        file_parts = file.split('_')
        metadata = {
            'path': os.path.join(root_abspath, file),
            'img_id': '_'.join(file_parts[3:5]) ,
            'case_id': int(file_parts[1][4:]),
            'slide_type': file_parts[0],
            'target': os.path.basename(root),
            'biomarker': file_parts[2],
            'tile_row_offset':  int(file_parts[7].split('.')[0][3:]),
            'tile_col_offset':  int(file_parts[6][3:])
        }
        metadata_list.append(metadata)
```

```python
df_metadata = pd.DataFrame(metadata_list)
```

# Adding original image path to metadata

```python
df_metadata = pd.read_parquet('../data/02_intermediate/spatial-attention-pcr/metadata.parquet')
```

```python
orig_image_paths = []
for i, metadata in tqdm(df_metadata.iterrows(), total=len(df_metadata)):
    folder_path = os.path.join(
        os.path.abspath('../data/01_raw/spatial-attention-pcr/'),
        metadata['target'],
        'openSlide_level_4_To_level_0_rigid_reg',
        f"CASE{metadata['case_id']}",
        metadata['img_id'])
    img_name = f"{metadata['slide_type']}_CASE{metadata['case_id']}_{metadata['biomarker']}_{metadata['img_id']}.jpg"
    orig_image_paths.append(os.path.join(folder_path, img_name))
```

```python
df_metadata['orig_img_path'] = orig_image_paths
```

```python
df_metadata.to_parquet('../data/02_intermediate/spatial-attention-pcr/metadata.parquet')
```
