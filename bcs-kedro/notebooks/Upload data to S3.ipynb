# Upload Data

```python
# Substitute <project_root> with the [root folder for your project](https://docs.kedro.org/en/stable/tutorial/spaceflights_tutorial.html#terminology)
conf_path = str(Path('../') / settings.CONF_SOURCE)
conf_loader = OmegaConfigLoader(conf_source=conf_path)
credentials = conf_loader["credentials"]
```

```python
s3_client = boto3.client(
    's3', 
    aws_access_key_id=credentials['dev_s3']['aws_access_key_id'], 
    aws_secret_access_key=credentials['dev_s3']['aws_secret_access_key'], 
    region_name='us-east-1'
)
```

```python
def object_exists(s3_client, bucket, object_path):
    try:
        s3_client.head_object(Bucket=bucket, Key=object_path)
    except botocore.exceptions.ClientError as e:
        return False
    else:
        return True
```

```python
DATA_FOLDER = r'D:\Dados mestrado\pcr\openSlide_level_4_To_level_0_rigid_reg'
```

```python
for root, dirs, files in os.walk(DATA_FOLDER):
    for file in files:
        local_file_path = Path(os.path.join(root, file))
        s3_object_path = Path(os.path.join('spatial-attention-pcr', *local_file_path.parts[2:])).as_posix()
        if not object_exists(s3_client, BUCKET, s3_object_path):
            print(f'Uploading file: {local_file_path}')pomodopomodo
            s3_client.upload_file(local_file_path, BUCKET, s3_object_path)
```
